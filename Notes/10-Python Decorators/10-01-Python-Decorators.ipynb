### 1. DEKORATÖRÜN (FABRİKANIN) TANIMLANMASI

```python
def new_decorator(original_func):
    def wrap_func():
        print('Some extra code, before the original function')
    
        original_func()
    
        print('Some extra code, after the original function!')
    return wrap_func
```

### 2. DEKORATÖRÜN KULLANILMASI (@ SYNTAX)

```python
# Python @ ifadesini gördüğünde altındaki fonksiyonu alır ve dekoratöre parametre olarak atar.
# Arka planda şu komut çalışır: 
# func_needs_decorator = new_decorator(func_needs_decorator)
@new_decorator
def func_needs_decorator():
    print("I want to be decorated!!")
```

### 3. FONKSİYONUN ÇAĞRILMASI

```python
# Sen ana fonksiyonu çağırdığını sanırsın ama aslında çalışan şey 'wrap_func'tır.
func_needs_decorator()
```

Dekoratörler, orijinal fonksiyonun kaynak koduna hiç dokunmadan ona yeni yetenekler (öncesinde veya sonrasında çalışan kodlar) eklemek için kullanılır.

"Açma/kapama" mantığı ve kullanım amaçları şunlardır:
* **Kolay Açma/Kapama:** Fonksiyona eklediğin özelliği iptal etmek istediğinde fonksiyonun içini silip düzenlemek yerine sadece başındaki @decorator satırını siler veya yoruma alırsın (# @decorator). İhtiyaç olduğunda tekrar aktif edersin.
* **Kod Tekrarını Önleme (DRY):** Onlarca farklı fonksiyonda aynı kontrolü veya işlemi yapman gerekiyorsa, bu mantığı tek bir dekoratörde toplayıp hepsinin başına @ ile eklersin.
* **Temel Kullanım Alanları:**
  * **Kimlik Doğrulama (Auth):** Kullanıcı giriş yapmış mı kontrol etmek (Web framework'lerinde sıkça: `@login_required`).
  * **Süre Ölçümü / Performans:** Fonksiyonun kaç saniyede çalıştığını loglamak.
  * **Loglama:** Fonksiyonun ne zaman, hangi argümanlarla çağrıldığını kaydetmek.
  * **Hata Yönetimi / Rate Limiting:** Fonksiyonun çalışma sıklığını sınırlamak.

# Aynı kodun yorumlu hali: 

```python
# new_decorator, dışarıdan bir fonksiyonu (original_func) parametre olarak alır.
def new_decorator(original_func):
    # İçeride sarmalayıcı (wrapper) yeni bir fonksiyon tanımlanır.
    # Ekstra kodlar ve orijinal fonksiyonun ne zaman çalışacağı burada belirlenir.
    def wrap_func():
        # Orijinal fonksiyondan ÖNCE çalışacak kodlar
        print('Some extra code, before the original function')
        
        # Parametre olarak gelen orijinal fonksiyonun çağrılması
        original_func()
        
        # Orijinal fonksiyondan SONRA çalışacak kodlar
        print('Some extra code, after the original function!')
        
    # new_decorator, oluşturduğu bu yeni fonksiyonu (wrap_func) ÇALIŞTIRMADAN döndürür.
    # Fonksiyonun referansını döndürüyoruz, yanına () koymuyoruz.
    return wrap_func
```

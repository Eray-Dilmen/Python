```python
import re
pattern = r'\d{3}-\d{3}-\d{4}'
```

```python
import os

for folder, sub_folders, files in os.walk("extracted files"):
    for file in files:
        
        file_path = os.path.join(folder, file)
        with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
            text = f.read()
            match = re.search(pattern, text) #pattern'i dosya içerisnde arayacak, dosya da text değişkeninde tutuyoruz
        if match:
            print(f"Phone: {match.group()}")
            print(f"Location: {file_path}")
```

### 📌 Not: os.path.join() Kullanımı ve Amacı

`os.path.join(path1, path2, ...)` fonksiyonu, verilen klasör yollarını ve dosya adlarını işletim sistemine uygun ayraçla (`/` veya `\`) birleştirerek **geçerli bir tam dosya yolu** oluşturur. (Yani parametre olarak dosyanın bulunduğu konumu ve dosyanın adını alırız).

* **1. Parametre (`folder`):** Dosyanın bulunduğu o anki klasörün yoludur (Örn: `'extracted files/extracted_content/Four'`).
* **2. Parametre (`file`):** Klasörün içindeki dosyanın adıdır (Örn: `'EMTGPSXQEJX.txt'`).

**Neden Gerekli?**

`files` listesinden gelen `file` değişkeni sadece dosyanın adını (`'EMTGPSXQEJX.txt'`) tutar. Doğrudan `open('EMTGPSXQEJX.txt')` yazarsan, Python o dosyayı alt klasörde değil, en üstteki ana çalışma dizininde (Jupyter'in çalıştığı dizinde) arar ve `FileNotFoundError` hatası verir. 

`os.path.join(folder, file)` kullandığında iki metni araya işletim sistemi ayracını koyarak birleştirir:

`'extracted files/extracted_content/Four'` + `'EMTGPSXQEJX.txt'` -> `'extracted files/extracted_content/Four/EMTGPSXQEJX.txt'`

Böylece `open()` fonksiyonu dosyanın tam konumuna ulaşıp dosyayı hatasız şekilde açabilir.

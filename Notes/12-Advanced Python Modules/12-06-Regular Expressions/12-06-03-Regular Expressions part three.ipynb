## Additional Regex Syntax

### Or operator |

Use the pipe operator to have an **or** statment. For example

```python
# iki veya daha fazla şey aradığımız zaman kullanıyoruz 

import re 

re.search(r'cat | dog', 'cat and dog') #tabiki ilkini gösteriyor 

for word in re.finditer(r'cat|dog', 'cat and dog'):
    print(word.group())
```

```python
# loopa aldığımız zaman ikisinin de bastırıldığını görüyoruz
```

### The Wildcard Character

Use a "wildcard" as a placement that will match any character placed there. You can use a simple period **.** for this. For example:

```python
re.findall(r'.at', 'The cat in the hat went splat')
```

```python
# ama daha fazla wildcard istiyorsak mesela splat için
```

```python
re.findall(r'...at', 'The cat in the hat went splat')
```

```python
# diğerlerinde de sonuç hatalı dönecektir bu göz önünde bulundurulamalıdır.
# önceki derslerde gördüğümüz filtreleme daha kullanışlıdır
```

### Starts with and Ends With

We can use the **^** to signal starts with, and the **$** to signal ends with:

```python
# mesela ^\d dediğimiz zaman sadece digit ile başlayana bakar 
# \d$ dediğimiz zaman sonu digit ile bitene bakar
```

```python
re.search(r'^\d','1 deneme') # başında 1 var 1 döndü
```

```python
re.search(r'^\d','deneme 1') # none döndü çünkü 1 ile başlamıyor
```

```python
re.search(r'\d$', 'deneme 1') #sonunda 1 var 1 döndü 
```

```python
re.search(r'\d$', '1 deneme') #sonunda digit yok none döndü 
```

### Exclusion

To exclude characters, we can use the **^** symbol in conjunction with a set of brackets **[]**. Anything inside the brackets is excluded. For example:

```python
text = "there are 3 numbers 45 of 9 this text."
```

```python
re.findall(r'[^\d]', text) # + sız hali:
```

---

```python
re.findall(r'[^\d]+', text) # + lı hali:
```

### `[^\d]` ile `[^\d]+` Arasındaki Fark (Neden Sonuç Değişti?)

Regex motoru varsayılan olarak **açgözlüdür (greedy)**; koşulu sağladığı sürece eşleşmeyi mümkün olan en uzun şekilde uzatır.

* **`[^\d]` (Artı işaretsiz):** "Rakam olmayan **tek bir** karakter bul."
  * Motor her karakteri bağımsız değerlendirir ve listeye tek tek atar: `'t'`, `'h'`, `'e'`, `' '`...

* **`[^\d]+` (Artı işaretli):** "Rakam olmayan karakterleri bul ve **arka arkaya geldikleri sürece hepsini tek parça** topla."
  * Motor karakterleri biriktirerek ilerler; ta ki kuralı bozan bir **rakama** çarpana kadar. Rakamı görünce o ana kadarki bloğu (`'there are '`) tek bir eleman olarak listeye ekler.

---
We can use this to remove punctuation from a sentence.

```python
text = 'This is a string! But it has punctuation. How can we remove it?'
```

```python
re.findall(r'[^!.? ]+', text) # boşluğu da ekledik( kelime kelime bulmak istiyorsak mesela ) 
#sonrasında joinle kelimeleri birleştirebiliriz 
```

```python
' '.join(re.findall(r'[^!.? ]+', text))
```

## Brackets for Grouping

belli bir kısmı bulmak içinde kullanılabiliyor, - ile bağlanmış kelimeleri bulmak için:

```python
text = 'Only find the hypen-words in this sentence. But you do not know how long-ish they are'
```

```python
re.findall(r'[\w]+-[\w]+',text)
```

## Parenthesis for Multiple Options

or | ile birden çok arama yapabileceğimizi söylemiştik, bu onun biraz daha farklı versiyonu, kelimenin devamında ne  geleceğini belirleyip filtreliyoruz

```python
text = 'Hello, would you like some catfish?'
texttwo = "Hello, would you like to take a catnap?"
textthree = "Hello, have you seen this caterpillar?"
all_texts = text + " " + texttwo + " " + textthree
```

```python
for words in re.finditer(r'cat(fish|nap|erpillar)',all_texts): # parantez kullanarak cat ifadesinden hemen sonra ne geleceğini belirliyoruz (birden çok )
    print(words.group())
```

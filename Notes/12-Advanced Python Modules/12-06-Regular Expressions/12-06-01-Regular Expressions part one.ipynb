```python
text = "The agent's phone number is 408-555-1234. Call soon!, better phone up"
```

```python
import re # regular expression
```

```python
pattern = 'phone'
```

```python
re.search(pattern,text) # hangi kelime aranacak, nerde aranacak ? 
```

```python
search = re.search(pattern, text)
search.span() # direkt olarak index başlangıç ve sonunu veriyor
```

```python
search.group() # içerilen bilgiyi veriyor
```

```python
# tüm geöen kelimeleri bulmak için 'findall' kullanıyoruz
```

```python
re.findall(pattern,text)
```

```python
# eğer aranan kelimenin tüm geçtiği yerleri bulmak istiyorsak 'finditer' kullanıyoruz

for words in re.finditer(pattern,text):
    print(words)
```

```python
# direkt olarak yerleri isteniyorsa

for words in re.finditer(pattern,text):
    print(words.span())
```

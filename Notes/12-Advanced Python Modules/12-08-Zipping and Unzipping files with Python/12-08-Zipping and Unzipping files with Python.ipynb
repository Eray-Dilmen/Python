öncelikle ziplenmiş bir dosya oluşturacağız, daha sonrasında bu ziplenmiş dosyanın içerisine atacağız

```python
comp_file.close()
```

## Extracting from Zip Files

**extractall()** diyerek içerisindeki tüm dosyaları çekebilir veya spesifik olarak almak istediğimiz dosyayı **extract()** ile alırız

```python
zip_obj = zipfile.ZipFile('comp_file.zip','r') # bu sefer w modu ile değil r modu ile yapıyoruz, içinden çekeceğiz
# ama current directoryde olduğumuz için direkt olarak ismini yazabildik yoksa path belirtmek zorundaydık 
# örnek -> r'/Users/.../comp_file.zip'
```

```python
zip_obj.extractall("extracted_content") # current file'da extracted_content oluşturup onun içine atmasını istiyoruz
```

## Using shutil library

Sadece dosya dosya çıkarıp ziplemek yerine folder'in tamamını ziplemek için **shutil** kütüphanesi kullanacağız

```python
import shutil
```

```python
#ziplenecek kaynak dosyası: 12-08 klasörünü veriyorum
directory_to_zip = '/Users/eraydilmen/Documents/12-Advanced Python Modules/12-08-Zipping and Unzipping files with Python'
```

```python
# Creating a zip archive
# Bu da oluşturulacak zip dosyasının lokasyonu (/12-08-ALL klasörümüz olmadığı için otomatik oluşturulacak)
output_filename = '/Users/eraydilmen/Documents/12-Advanced Python Modules/12-08-ALL'
shutil.make_archive(output_filename,'zip',directory_to_zip) # zip'in adı ne olacak, formatı ne, path'i ne 
```

```python
# Extracting a zip archive
# folder ismini 12-08-EXTRACTED verdik ki tüm dışarı çıkarılıcak dosyaları oraya atsın
output_filename = '/Users/eraydilmen/Documents/12-Advanced Python Modules/12-08-ALL.zip'
dir_for_extract_result = '/Users/eraydilmen/Documents/12-Advanced Python Modules/12-08-EXTRACTED' 
shutil.unpack_archive(output_filename, dir_for_extract_result,'zip') 

# nerden çıkarılacak (dosyanın adını yazarken uzantısını da yazıyoruz '.zip') ,
# nereye çıkarılacak ,
# current dosyanın formatı ne
```

### ⚠️ Not: Zip ve Çıkarma İşlemlerinin Üst Dizinde Yapılma Sebebi (Recursion / Bozulma Bug'ı)

Bu çalışmada oluşturulan `12-08-ALL.zip` arşivi ve `12-08-EXTRACTED` klasörü bilerek bir üst dizinde (`12-Advanced Python Modules`) oluşturulmuştur:

* **Kendini Kendi İçine Zipleme Sorunu (Recursion):** Eğer kaynak klasörün kendisini (`12-08...`) ziplerken çıktı zip dosyasını da yine aynı klasörün içine kaydedersek; Python ziplenecek dosyaları okurken o sırada yazmakta olduğu `.zip` dosyasını da kaynak veri sanıp arşivin içine eklemeye çalışır. Bu durum dosyanın bozulmasına ve arşiv açılırken `ReadError` / `is not a zip file` hatalarına yol açar.
* **Çözüm:** Ziplenecek kaynak klasör ile hedef çıktı yolunu birbirinden ayırmak ve zip/extract işlemlerini kaynak dizinin **dışında (üst klasörde)** gerçekleştirmektir.

`pdb` (Python Debugger), Python'ın yerleşik **kod hata ayıklama (debugging)** aracıdır.

Sürekli `print()` yazarak değişkenlerin değerini kontrol etmek yerine, kodun akışını istediğin satırda **durdurmanı**, o anki hafızayı incelemeni ve satır satır ilerlemeni sağlar.

### Temel Görevleri

* Kodun içine durma noktası (**breakpoint**) ekler.
* Program durduğunda terminalde etkileşimli bir kabuk açar; değişkenlerin anlık değerlerini sorgulamana veya değiştirmene izin verir.
* Hatanın tam olarak hangi satırda, hangi değişken değeri yüzünden patladığını anında görmeni sağlar.

### Kod Örneği

Aşağıdaki kodda bir toplama işlemi hatası simüle edelim:

```python
import pdb

x = [1, 3, 4]
y = 2
z = 3

result_one = y + z

# Programı burada durduruyoruz
pdb.set_trace()

# Hatalı satır
result_two = y + x
```

### Nasıl Kullanılır? (Terminal / Hücre Arayüzü)

Kod `pdb.set_trace()` satırına geldiğinde çalışmayı durdurur ve `(Pdb)` istemcisi açılır.

Kullanabileceğin temel komutlar:

* **Değişken Adı:** O anki değerini görmek için doğrudan adını yazıp Enter'a basabilirsin (ör. `x`, `result_one`).
* **`q` (quit):** Hata ayıklayıcıdan çıkar ve programı tamamen sonlandırır.
* **`c` (continue):** Bir sonraki breakpointe kadar veya program bitene kadar devam eder.
* **`n` (next):** Bir sonraki satıra geçer.

## default dictionary

dict'te olmayan bir key'in valuesini aramaya calıstığımızda hata verir,
fakat defaultdict kullanırsak verdiğimiz default değerini yeni oluşturduğumuz key'e otomatik atar.

```python
d = {'a' : 10}
d['a']
```

```python
d['correct']
```

```python
from collections import defaultdict

d = defaultdict(lambda: 0)

d['correct'] = 100
d['correct']
```

```python
d['wrong']
```

```python
dict(d)
```

### `namedtuple` Nedir?

Normal tuple'larda elemanlara yalnızca indeksleri (`t[0]`, `t[1]`) ile erişilebilir. **`namedtuple`**, tuple'ın hafifliğini ve değiştirilemezlik (immutable) özelliğini korurken, her alana bir **isim** vererek elemanlara nesne özniteliği gibi (`sammy.age`, `sammy.name`) erişmeyi sağlar. Basit sınıflar (class) tanımlamak yerine pratik bir alternatiftir.

```python
from collections import namedtuple

# 1. namedtuple Şablonunu (Sınıfını) Tanımlama
# Parametreler: (Tip Adı, [Alan İsimleri])
Dog = namedtuple('Dog', ['age', 'breed', 'name'])

# 2. Nesne Oluşturma
sammy = Dog(age=5, breed='Husky', name='Sam')

# 3. Tipini Kontrol Etme
type(sammy)
# Çıktı: __main__.Dog

# 4. Nesneyi Görüntüleme
sammy
# Çıktı: Dog(age=5, breed='Husky', name='Sam')
```

```python
# 5. Elemanlara Erişim Yöntemleri

# İsim ile erişim (En yaygın kullanım):
print(sammy.age)    # 5
print(sammy.breed)  # Husky
print(sammy.name)   # Sam

# Normal tuple gibi indeks ile erişim:
print(sammy[0])     # 5
print(sammy[1])     # Husky
```

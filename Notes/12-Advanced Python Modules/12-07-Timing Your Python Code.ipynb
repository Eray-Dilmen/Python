```python
def func_one(n):
    '''
    Given a number n, returns a list of string integers
    ['0','1','2',...'n]
    '''
    return [str(num) for num in range(n)]

func_one(10)
```

---

```python
def func_two(n):
    '''
    Given a number n, returns a list of string integers
    ['0','1','2',...'n]
    '''
    return list(map(str,range(n)))

func_two(10)
```

```python
import time
```

#### 1.fonksiyon için yapalım 

```python
# STEP 1: Get start time
start_time = time.time()
# Step 2: Run your code you want to time
result = func_one(1000000)
# Step 3: Calculate total time elapsed
end_time = time.time() - start_time

end_time
```

#### 2.fonksiyon için:

```python
# STEP 1: Get start time
start_time = time.time()
# Step 2: Run your code you want to time
result = func_two(1000000)
# Step 3: Calculate total time elapsed
end_time = time.time() - start_time

end_time
```

### 2.yöntem: Timeit modülü

```python
import timeit # builtin geliyor zaten istediğimiz gibi kullanabiliriz
```

 timeit.timeit( ) burda parametre içi 3 tane değer alıyor. 
 
 ilki neyi test etmek istediğimiz (statement)
 
 ikincisi neyi test edeceğimiz (setup) yani fonksiyonun kendisini veriyoruz
 
 üçüncüsü sayı, kaç defa çalışacağını belirliyoruz

 ### `timeit.timeit()` Parametre Sıralaması Notu

`timeit.timeit(stmt, setup, timer, number, globals)` imzasında `number` 4. parametredir (3. sırada `timer` bulunur). Bu yüzden `number` değerini verirken `number=100000` şeklinde anahtar kelime (keyword argument) olarak yazmak gerekir. Doğrudan 3. sıraya sayı yazılırsa `timer` parametresine atanacağı için hata verir.

```python
stmt = 'func_one(100)'

setup = '''
def func_one(n):
    return [str(num) for num in range(n)]
'''
```

```python
timeit.timeit(stmt, setup, number=100000) # üçüncüyü de number olarak belirtmek zorundayız
```

```python
stmt2 = 'func_two(100)'

setup2 = '''
def func_two(n):
    return list(map(str,range(n)))
'''
```

```python
timeit.timeit(stmt2,setup2,number=100000)
```

#### 2.fonksiyon daha hızlı çalıştı, daha iyi görmek için parametre seviyesini arttıralım 

---

```python
timeit.timeit(stmt,setup,number=1000000) #1.fonksiyon için çalıştırıyoruz
```

```python
timeit.timeit(stmt2,setup2,number=1000000) # 2.fonksiyon için çalıştırıyoruz
```

```python
# yaklaşık 700 milisaniye fark ile 2.fonksiyon daha hızlı çalışıyor
```

### Aritmetik olarak iki tarih arasındaki farkı hesaplama

```python
from datetime import date #date object'ini import ettik
```

```python
date1 = date(2021,8,20)
date2 = date(2022,8,21)

result = date2-date1
result # gün olarak return ediyor
```

```python
result.days
```

```python
# ya da direkt olarak
print(result)
```

### şimdi datetime object'ine bakalım 

```python
from datetime import datetime
datetime1 = datetime(2021,8,20,5,0)
datetime2 = datetime(2022,8,20,6,0)

result_d = datetime2-datetime1 # hem gün hem saniye olarak return ediyor
result_d
```

```python
# 1 gün fark 3600 saniye
```

```python
result_d.seconds
```

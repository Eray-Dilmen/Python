### random.seed() fonksiyonu
Yaptığı tek şey, üretilecek sayı serisinin başlangıç noktasını sabitlemektir.

### Çalışma Mantığı

* **`random.seed(x)` çağrılmadığında:** Python başlangıç noktası olarak genellikle anlık sistem saatini alır. Bu yüzden her çalıştırmada farklı sayılar üretilir.
* **`random.seed(101)` yazıldığında:** Algoritmanın başlangıç indeksi sabit bir noktaya çekilir. Aynı seed değeri verildikten sonra üretilen sayı dizisi **her zaman aynı sırayla** çıkar.

```python
import random

# 1. Tur: Seed veriyoruz
random.seed(101)  #içerisindeki parametre arbitrary(keyfi)
print(random.randint(0, 100))  # 74
print(random.randint(0, 100))  # 24
print(random.randint(0, 100))  # 69
```

```python
# 2. Tur: Tekrar aynı seed'i verip başa sarıyoruz
random.seed(101)
print(random.randint(0, 100))  # Yine 74
print(random.randint(0, 100))  # Yine 24
print(random.randint(0, 100))  # Yine 69
```

### Örnekleme Yöntemleri (Sampling)

* **`random.choices()` (Sample with Replacement / Yerine Koyarak Örnekleme):** Seçilen eleman listeye geri bırakılır. Bu yüzden aynı eleman birden fazla kez seçilebilir (çıktıda tekrar eden sayılar görülebilir).
* **`random.sample()` (Sample without Replacement / Yerine Koymadan Örnekleme):** Seçilen eleman listeden çıkarılmış gibi davranılır. Bu yüzden aynı eleman ikinci kez seçilemez (tüm elemanlar benzersizdir).

```python
# SAMPLE WITH REPLACEMENT
random.choices(population=mylist, k=10) # isimlerini açıkça yazmak zorunlu değildir. (mylist, 10) da yazabiliriz
```

```python
# SAMPLE WITHOUT REPLACEMENT 
random.sample(population=mylist, k=10)
```

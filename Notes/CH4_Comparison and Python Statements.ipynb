### If, Elif, Else


```python
if 3<2:
    print("its true")
else:
    print("not true")
```

```python
loc = "bank"

if loc == "deneme":
    print("bank denemeye eşit")
elif loc == "bank":
    print("doğru çıktı")
else:
    print("salla")
```

### For Loops

```python
list1 = [1,2,3,4,5]
for a in list1:
    print(a)
```

```python
list2 = [1,2,3,4,5]
for b in list2:
    if b % 2 == 0:
        print(b)
    else:
        print(f"odd:{b}")
```

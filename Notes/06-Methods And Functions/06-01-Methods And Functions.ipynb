```python
# dizi içerisinde çift olan sayıları getiren fonksiyon

def check_even_list(num_list):
    even_numbers = []
    for number in num_list:
        if number % 2 == 0:
            even_numbers.append(number)
            return even_numbers
        else: pass

check_even_list([2, 4, 5])  
```

```python
mynums = [1, 2, 3, 4, 5, 6]

# check_even diye ayrı bir fonksiyon yazmadan direkt içine yazdık:
evens = list(filter(lambda num: num % 2 == 0, mynums))
evens
```

### Map 


```python
nums = [1,2,3,4,5]

def take_square(x):
    return x**2

list(map(take_square, nums)) ## fonksiyonu 2.parametre olarak verilen dizideki her bir eleman için uyguluyor.
```

### Filter
Filter ise sadece dizideki elemanlardan fonksiyonda True return edilenleri seçiyor

---
mesela dizi içerisinde sadece çift olan sayıları almak istiyoruz: 

```python
nums = [1,2,3,4,5,6]

def is_even(x):
    return x%2 == 0

list(filter(is_even,nums))
```

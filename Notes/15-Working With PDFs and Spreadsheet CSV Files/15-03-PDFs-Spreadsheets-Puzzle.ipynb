## Task One: Grab the Google Drive Link from .csv File

### CSV Matrisinde Çapraz (Diyagonal) Veri Çekme Mantığı

Metin, $N \times M$ boyutundaki bir matrisin sol üst köşesinden sağ alt köşesine doğru **çapraz (diyagonal)** olarak gizlenmiştir.

* **Çalışma Prensibi:**
  * `enumerate(data_lines)` fonksiyonu her adımda iki değer üretir: 
    * `row_index`: Satır numarası ($0, 1, 2, 3 \dots$)
    * `row`: O satırdaki tüm sütunların listesi
  * `row[row_index]` yazılarak satır numarası ile sütun indeksinin aynı olması sağlanır.

* **Örnek İndeks Adımları:**
  * $0$. satır, $0$. sütun $\rightarrow$ `data_lines[0][0]` $\rightarrow$ `'h'`
  * $1$. satır, $1$. sütun $\rightarrow$ `data_lines[1][1]` $\rightarrow$ `'t'`
  * $2$. satır, $2$. sütun $\rightarrow$ `data_lines[2][2]` $\rightarrow$ `'t'`
  * $3$. satır, $3$. sütun $\rightarrow$ `data_lines[3][3]` $\rightarrow$ `'p'`

> **Sözdizimi Kuralı:** `row` bir liste, `row_index` ise bir sayıdır. Bu nedenle erişim her zaman `row[row_index]` şeklinde olmalıdır;

## Task Two: Download the PDF from the Google Drive link and find the phone number that is in the document. 

```python
# You should get this phone number
# 505 503 4455
```

```python
import pypdf
phone_format = r'\d{3}.\d{3}.\d{4}' # . ifadesi wildcardtır \n harici herhangi bir şey gelebilir. araya ne geleceğini bilmediğimizden böyle yaptık
phone = ''
```

```python
f = open('Exercise_Files/Find_the_Phone_Number.pdf', 'rb') #dosyayı aldık şimdi okutacaz
pdf_reader = pypdf.PdfReader(f) #dosyayı okuyup iterasyon eden bu değişken artık 
len(pdf_reader.pages) # 17 sayfa varmış loopa sokarken 17 sayfa girmemiz gerektiğini biliyoruz
```

```python
import re
for page in pdf_reader.pages: # her sayfa için
    text = page.extract_text()
    match = re.search(phone_format, text) # phone_formatını arıyorum, nerde arıyorum text değişkeninde(o da page)
    if match:
        phone+=(match.group()) # veya print diyebilirdik veya listeye alıp phone.append diyebilirdik
```

```python
import pypdf
import re

# 1. PDF dosyasını aç ve tüm sayfaları tek bir metinde birleştir
f = open('Exercise_Files/Find_the_Phone_Number.pdf', 'rb')
pdf_reader = pypdf.PdfReader(f)

all_text = ''
for page in pdf_reader.pages:
    all_text += ' ' + page.extract_text()

f.close()

# 2. Geniş kapsamlı regex kalıbı (örn: 3 basamaklı tüm sayı grupları)
pattern = r'\d{3}'

# 3. Metindeki tüm eşleşmeleri ve konumlarını (span) listele
for match in re.finditer(pattern, all_text):
    print(match)
```

```python
# 4. Tespit edilen başlangıç indeksinden itibaren metni dilimleme (slicing)
# 42624 indeksinden başlayarak yaklaşık 12 karakterlik alanı çek
phone_area = all_text[42624:42624 + 12]
print(phone_area)
```

```python
# 5. Tam telefon numarası formatı için daraltılmış regex çalıştırma
phone_pattern = r'\d{3}.\d{3}.\d{4}'
match = re.search(phone_pattern, all_text)

if match:
    print("Bulunan Telefon Numarası:", match.group())
```

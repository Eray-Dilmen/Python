### tüm sayfaları bir dosyaya atmak için: (dosyamızın adı: working_business_created_copy.pdf )

```python
import pypdf

# 1. Orijinal dosyayı oku
f = open('Working_Business_Proposal.pdf', 'rb')
pdf_reader = pypdf.PdfReader(f)

# 2. Yeni bir PDF yazıcı nesnesi oluştur
pdf_writer = pypdf.PdfWriter()

# 3. Tüm sayfaları tek tek yeni PDF'e ekle
for page in pdf_reader.pages:
    pdf_writer.add_page(page)

# 4. Yeni dosyayı diske kaydet
pdf_output = open('working_business_created_copy.pdf', 'wb')
pdf_writer.write(pdf_output)

# 5. Dosyaları kapat
f.close()
pdf_output.close()
```

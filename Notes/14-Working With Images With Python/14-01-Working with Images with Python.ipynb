# Color Transparency

#### RGBA - Red, Green, Blue, Alpha

```python
# alpha değeri 0 ise foto tamamen görünmez oluyor, 255 ise tamamen opak oluyor
```

```python
red = Image.open('red_color.jpg')
red = red.convert('RGBA')
red
```

```python
blue = Image.open('blue_color.png')
```

```python
blue
```

```python
blue = blue.convert('RGB')
blue.putalpha(128)
blue
```

```python
blue.putalpha(20)
blue
```

```python
red.putalpha(128)
red
```

```python
blue.paste(im = red , box = (0,0), mask = red)
blue # 128 blue değerindeyken 128 red değerini üstüne yapıştırdık (blue 20 yi çalıştırmadım restartlayıp 128 üzerinden gittik)
```

```python
blue.save('purple.png')
purple = Image.open('purple.png')
purple # görüldüğü üzere purple olarak yeni oluşanı kaydettik ve dosyayı açtığımızda bu renk bizi karşılıyor.
```

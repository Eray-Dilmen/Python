```python
# Gerekli widget ve etkileşim bileşenlerini içe aktarma
from ipywidgets import interact, interactive, fixed
import ipywidgets as widgets
```

```python
# Girdi değerini aynen döndüren fonksiyon
def func(x):
    return x
```

```python
# Tamsayı (int) değeri verildiğinde otomatik olarak bir int slider (kaydırma çubuğu) üretir
interact(func, x=10)
```

---

```python
def func(x):
    return x**2
```

```python
interact(func, x=10) # yukarıdaki örneğin aynısının 2 ile çarpılmış hali, inputa bağlı olarak output üssü 2 olarak çıkacak 
```

---

```python
interact(func, x=True) # Checkbox olarak çıkıyor inputa göre sonuç 1 veya 0 oluyor
```

```python
def func(x):
    return x
```

```python
interact(func, x='Hello') # input olarak aşağıya ne yazarsak outputta senkronize şekilde o gözükecek
```

```python
# @interact dekoratörü: x için boolean (onay kutusu/checkbox), y için float (ondalıklı sayı kaydırma çubuğu/float slider) oluşturur
@interact(x=True, y=1.0)
def g(x, y):
    return (x, y)
```

```python
# yukarıdaki örneğin aynısı ama 'fixed' ile y parametresini sabit tutup widget oluşturmasını engeller
@interact(x=True, y=fixed(1.0))
def g(x, y):
    return (x, y)
```

---

```python
# IntSlider kullanarak minimum (-100), maksimum (100), adım aralığı (1) ve başlangıç değeri (0) özel olarak tanımlanmış tamsayı kaydırıcısı oluşturur
interact(func, x=widgets.IntSlider(min=-100, max=100, step=1, value=0))
```

---

```python
# IntSlider'ın kısayol (tuple) gösterimi: (min, max, step) -> (-10'dan 10'a, 1'er artışla)
interact(func, x=(-10, 10, 1))
```

---

```python
# FloatSlider için (min, max, step) kısayolu; fonksiyondaki varsayılan değer (x=5.0) slider'ın başlangıç konumunu belirler
@interact(x=(0.0, 20.0, 0.5))
def h(x=5.0):
    return x
```

---

```python
# Tuple () sayısal aralık belirtip Slider üretirken, List [] vermek açılır menü (Dropdown) oluşturur
interact(func, x=['hello', 'option 2', 'option3'])
```

---

```python
# Sözlük (dict) kullanımı: Anahtarlar (keys) menüdeki görünen etiketleri, değerler (values) fonksiyona aktarılan gerçek verileri temsil eder
interact(func, x={'one': 10, 'two': 20})
```

```python
# Görüntüleme modülünü içe aktarma ve iki sayıyı toplayıp ekrana basan fonksiyon
from IPython.display import display
```

```python
def f(a, b):
    display(a + b)
    return a + b
```

```python
# interactive(): interact'in aksine widget'ı hemen ekrana çizmez, daha sonra kullanılmak üzere bir nesne (widget) olarak döndürür
w = interactive(f, a=10, b=20)
```

```python
# Döndürülen 'w' nesnesinin tipini kontrol etme
type(w)
```

```python
# w nesnesinin alt bileşenlerini (slider'ları ve çıktı alanını içeren tuple) listeleme
w.children
```

```python
# interactive() ile oluşturulup 'w' değişkenine atanan widget nesnesini arayüzde görsel olarak ekrana basar
display(w)
```

```python
import ipywidgets as widgets

# Bağımsız bir IntSlider nesnesi oluşturup değişkene atıyoruz
# (Bir fonksiyona bağlamadan saf arayüz elemanı olarak üretir)
w = widgets.IntSlider()

from IPython.display import display

# (Avantajı: Widget'ı fonksiyonlardan bağımsız yönetmeyi, özelliklerini (w.value gibi) ayrı hücrelerde okuyup değiştirmeyi ve istenen yerde tekrar tekrar göstermeyi sağlar)
display(w)
```

```python
display(w) 
# Senkronizasyon sebebi: İki görünüm de bellekteki aynı nesneye (w) bağlıdır; 
#arka plandaki model tek olduğu için birindeki değişiklik diğerine anında yansır
```

```python
w.close()
```

```python
# IntSlider widget nesnesi oluşturma ve ekranda gösterme
w = widgets.IntSlider()
display(w)
```

```python
# w.value: Slider'ın o anki güncel değerini (current value) Python tarafında okur
# Not: Slider hareket ettirildiğinde bu hücredeki çıktı otomatik güncellenmez; yeni değeri görmek için bu hücreyi tekrar çalıştırmak gerekir
w.value
```

```python
# bu atama yapıldığında arayüzdeki slider da anında 50 konumuna güncellenir
w.value = 50
```

```python
# w.keys: Widget nesnesinin sahip olduğu tüm dinamik ve senkronize edilebilir niteliklerin (trait attributes) isim listesini döndürür
# Örneğin; 'value', 'max', 'min', 'step', 'disabled', 'orientation' gibi dışarıdan erişilip değiştirilebilen tüm parametreleri görmeyi sağlar
w.keys
```

```python
#Slider widget'ının alabileceği üst sınırı (maksimum değeri) dinamik olarak 2000 yapar;
#arayüzdeki çubuk yeniden çizilmeden bu yeni limite göre güncellenir
w.max = 2000
```

---

```python
# Sayısal metin kutusu (FloatText) ve ondalıklı kaydırma çubuğu (FloatSlider) oluşturma
a = widgets.FloatText()
b = widgets.FloatSlider()

# İki widget'ı ekranda alt alta görüntüleme
display(a, b)

# widgets.jslink: İki widget'ın 'value' özelliğini tarayıcı tarafında (JavaScript ile) çift yönlü birbirine bağlar
# Birinden yapılan değişiklik (kutudan yazmak veya slider'ı kaydırmak) diğerini anında ve otomatik günceller
mylink = widgets.jslink((a, 'value'), (b, 'value'))
```

---

```python
# Sayısal metin kutusu ve kaydırma çubuğu tanımlama
a = widgets.FloatText()
b = widgets.FloatSlider()

# Widget'ları ekranda gösterme
display(a, b)

# Değişiklik: (b, 'value') yerine (b, 'max') bağlandı.
# Etkisi: Metin kutusuna girilen sayı (a.value), slider'ın anlık değerini değil maksimum sınırını (b.max) belirler. 
# Sonuç: Kutudan 100 yazıldığında slider'ın üst limiti 100 olur; slider artık 0 ile kutuya girilen sayı arasında dinamik olarak kaydırılabilir.
mylink = widgets.jslink((a, 'value'), (b, 'max'))
```

```python
# mylink.unlink(): İki widget arasındaki jslink bağlantısını koparır/iptal eder
# Bu komut çalıştırıldıktan sonra widget'lar tamamen bağımsız hale gelir; birinde yapılan değişiklik diğerini etkilemez
mylink.unlink()
```

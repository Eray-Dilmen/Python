# Widget List

This lecture will serve as a reference for widgets, providing a list of the GUI widgets available!

```python
import ipywidgets as widgets

# Show all available widgets
for item in widgets.Widget.widget_types.items():
    print(item[0][2][:-5])
```

## Numeric widgets

There are 10 widgets distributed with IPython that are designed to display numeric values. Widgets exist for displaying integers and floats, both bounded and unbounded.  The integer widgets share a similar naming scheme to their floating point counterparts. By replacing `Float` with `Int` in the widget name, you can find the Integer equivalent.

### IntSlider

```python
#Belirli bir aralıkta tek bir tam sayı değeri seçmek için yatay kaydırma çubuğu oluşturur.
widgets.IntSlider(
    value=7,
    min=0,
    max=10,
    step=1,
    description='Test:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True,
    readout_format='d'
)
```

### FloatSlider

```python
#IntSlider'dan farklı olarak ondalıklı (float) sayılarla çalışır ve hassas sayısal seçim sağlar.
widgets.FloatSlider(
    value=7.5,
    min=0,
    max=10.0,
    step=0.1,
    description='Test:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True,
    readout_format='.1f',
)
```

```python
#FloatSlider (Vertical): orientation='vertical' parametresiyle slider çubuğunu yatay yerine dikey olarak konumlandırır.
widgets.FloatSlider(
    value=7.5,
    min=0,
    max=10.0,
    step=0.1,
    description='Test:',
    disabled=False,
    continuous_update=False,
    orientation='vertical',
    readout=True,
    readout_format='.1f',
)
```

### IntRangeSlider

```python
# Belirli bir aralıkta iki uç nokta seçerek [min, max] şeklinde tam sayı aralığı belirlemeyi sağlayan çift başlıklı kaydırma çubuğudur.
widgets.IntRangeSlider(
    value=[5, 7],
    min=0,
    max=10,
    step=1,
    description='Test:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True,
    readout_format='d',
)
```

### FloatRangeSlider

```python
# IntRangeSlider'dan farklı olarak tam sayılar yerine ondalıklı (float) sayılarla bir değer aralığı ([min, max]) seçmek için kullanılır.
widgets.FloatRangeSlider(
    value=[5, 7.5],
    min=0,
    max=10.0,
    step=0.1,
    description='Test:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True,
    readout_format='.1f',
)
```

### IntProgress

```python
# Bir işlemin ilerleme durumunu göstermek için tam sayı tabanlı görsel bir ilerleme çubuğu (progress bar) oluşturur.
widgets.IntProgress(
    value=7,
    min=0,
    max=10,
    step=1,
    description='Loading:',
    bar_style='', # 'success', 'info', 'warning', 'danger' or ''
    orientation='horizontal'
)
```

### FloatProgress

```python
# IntProgress'ten farklı olarak ilerleme yüzdesini ondalıklı (float) hassasiyetle gösterir; bar_style ile renk teması belirlenebilir.
widgets.FloatProgress(
    value=7.5,
    min=0,
    max=10.0,
    step=0.1,
    description='Loading:',
    bar_style='info',
    orientation='horizontal'
)
```

The numerical text boxes that impose some limit on the data (range, integer-only) impose that restriction when the user presses enter.

### BoundedIntText

```python
# Kullanıcının sadece belirli sınırlar (min, max) arasında tam sayı girebildiği, klavye ve artırma/azaltma oklarıyla kontrol edilen metin kutusudur.
widgets.BoundedIntText(
    value=7,
    min=0,
    max=10,
    step=1,
    description='Text:',
    disabled=False
)
```

### BoundedFloatText

```python
# BoundedIntText'ten farklı olarak belirlenen alt-üst sınırlar dahilinde ondalıklı (float) sayı girişine izin veren sınırlı metin kutusudur.
widgets.BoundedFloatText(
    value=7.5,
    min=0,
    max=10.0,
    step=0.1,
    description='Text:',
    disabled=False
)
```

### IntText

```python
# BoundedIntText'ten farklı olarak herhangi bir alt/üst sınır (min, max) kısıtlaması olmadan serbest tam sayı girişi sağlayan metin kutusudur.
widgets.IntText(
    value=7,
    description='Any:',
    disabled=False
)
```

### FloatText

```python
# BoundedFloatText'ten farklı olarak alt/üst limit kısıtlaması bulunmayan, serbest ondalıklı sayı girişi yapılan kutudur.
widgets.FloatText(
    value=7.5,
    description='Any:',
    disabled=False
)
```

## Boolean widgets

There are three widgets that are designed to display a boolean value.

### ToggleButton

```python
# Tıklandığında basılı veya basılmamış durumda kalan (True/False), iki durumlu açma/kapama (anahtar) butonu oluşturur.
widgets.ToggleButton(
    value=False,
    description='Click me',
    disabled=False,
    button_style='', # 'success', 'info', 'warning', 'danger' or ''
    tooltip='Description',
    icon='check'
)
```

### Checkbox

```python
# ToggleButton gibi boolean (True/False) çalışan, ancak buton yerine standart bir onay kutusu arayüzü sunan bileşendir.
widgets.Checkbox(
    value=False,
    description='Check me',
    disabled=False
)
```

### Valid

The valid widget provides a read-only indicator.

```python
# Kullanıcı girdisi almayan, sadece bir işlemin veya girdinin doğruluğunu/geçerliliğini göstermek için kullanılan 
#salt okunur (read-only) durum göstergesidir (yeşil tik veya kırmızı çarpı).
widgets.Valid(
    value=False,
    description='Valid!',
)
```

## Selection widgets

There are several widgets that can be used to display single selection lists, and two that can be used to select multiple values. All inherit from the same base class. You can specify the **enumeration of selectable options by passing a list** (options are either (label, value) pairs, or simply values for which the labels are derived by calling `str`). You can **also specify the enumeration as a dictionary**, in which case the **keys will be used as the item displayed** in the list and the corresponding **value will be used** when an item is selected (in this case, since dictionaries are unordered, the displayed order of items in the widget is unspecified).

### Dropdown

```python
# Dropdown (Liste ile): Verilen liste elemanlarından kullanıcının tek bir seçenek seçmesini sağlayan açılır menüdür.
widgets.Dropdown(
    options=['1', '2', '3'],
    value='2',
    description='Number:',
    disabled=False,
)
```

The following is also valid:

```python
# Dropdown gibi tek bir seçenek seçtirir; ancak açılır menü yerine seçenekleri dikey bir liste kutusu (listbox) içinde gösterir.
widgets.Dropdown(
    options={'One': 1, 'Two': 2, 'Three': 3},
    value=2,
    description='Number:',
)
```

### RadioButtons

```python
# Dropdown'dan farklı olarak tüm seçenekleri açılır menüye gizlemeden açıkça listeleyen ve listeden yalnızca tek bir seçenek işaretlenmesine izin veren radyo butonu grubudur.
widgets.RadioButtons(
    options=['pepperoni', 'pineapple', 'anchovies'],
    # value='pineapple',
    description='Pizza topping:',
    disabled=False
)
```

### Select

```python
# Dropdown gibi tek bir seçenek seçtirir; ancak açılır menü yerine seçenekleri dikey bir liste kutusu (listbox) içinde gösterir.
widgets.Select(
    options=['Linux', 'Windows', 'OSX'],
    value='OSX',
    # rows=10,
    description='OS:',
    disabled=False
)
```

### SelectionSlider

```python
# Sayısal değerler yerine metinsel veya özel liste elemanları arasında kaydırma çubuğu (slider) ile tek bir eleman seçilmesini sağlar.
widgets.SelectionSlider(
    options=['scrambled', 'sunny side up', 'poached', 'over easy'],
    value='sunny side up',
    description='I like my eggs ...',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True
)
```

### SelectionRangeSlider
The value, index, and label keys are 2-tuples of the min and max values selected. The options must be nonempty.

```python
#SelectionSlider'dan farklı olarak, metinsel veya özel elemanlar (örneğin aylar/tarihler) içeren bir listeden (başlangıç, bitiş) şeklinde bir eleman aralığı seçmek için çift başlıklı slider sunar.
import datetime
dates = [datetime.date(2015,i,1) for i in range(1,13)]
options = [(i.strftime('%b'), i) for i in dates]
widgets.SelectionRangeSlider(
    options=options,
    index=(0,11),
    description='Months (2015)',
    disabled=False
)
```

### ToggleButtons

```python
# ToggleButton'ın tekli yapısından farklı olarak, yan yana dizilmiş butonlar arasından yalnızca bir tanesinin basılı/seçili kalmasını sağlayan buton grubudur.
widgets.ToggleButtons(
    options=['Slow', 'Regular', 'Fast'],
    description='Speed:',
    disabled=False,
    button_style='', # 'success', 'info', 'warning', 'danger' or ''
    tooltips=['Description of slow', 'Description of regular', 'Description of fast'],
    # icons=['check'] * 3
)
```

### SelectMultiple
Multiple values can be selected with <kbd>shift</kbd> and/or <kbd>ctrl</kbd> (or <kbd>command</kbd>) pressed and mouse clicks or arrow keys.

```python
# Select bileşeninden farklı olarak, listeden klavye tuşları (Shift, Ctrl / Cmd) veya fare ile birden fazla eleman seçilmesine izin veren liste kutusudur.
widgets.SelectMultiple(
    options=['Apples', 'Oranges', 'Pears'],
    value=['Oranges'],
    # rows=10,
    description='Fruits',
    disabled=False
)
```

## String widgets
There are several widgets that can be used to display a string value. The `Text` and `Textarea` widgets accept input. The `HTML` and `HTMLMath` widgets display a string as HTML (`HTMLMath` also renders math). The `Label` widget can be used to construct a custom control label.

### Text

```python
# Kullanıcıdan tek satırlık kısa metin (string) girişi almak için kullanılan temel metin kutusudur.
widgets.Text(
    value='Hello World',
    placeholder='Type something',
    description='String:',
    disabled=False
)
```

### Textarea

```python
# Text bileşeninden farklı olarak paragraf, çok satırlı metin veya kod girişi yapılabilen geniş metin alanıdır
widgets.Textarea(
    value='Hello World',
    placeholder='Type something',
    description='String:',
    disabled=False
)
```

### Label
The `Label` widget is useful if you need to build a custom description next to a control using similar styling to the built-in control descriptions.

```python
# Yanındaki widget'lara standart başlıklar yerine matematiksel ifadeler (LaTeX) veya özel metinler eklemek için kullanılan sabit metin/etiket alanıdır.
widgets.HBox([widgets.Label(value="The $m$ in $E=mc^2$:"), widgets.FloatSlider()])
```

### HTML

```python
# Standart HTML etiketlerini (<b>, <i>, <p> vb.) kullanarak zengin biçimli metinler ve stilize içerikler render etmeyi sağlar.
widgets.HTML(
    value="Hello <b>World</b>",
    placeholder='Some HTML',
    description='Some HTML',
)
```

### HTML Math

```python
# HTML bileşeninden farklı olarak hem HTML etiketlerini hem de MathJax/LaTeX formatındaki matematiksel formülleri doğrudan yorumlayıp ekrana basar.
widgets.HTMLMath(
    value=r"Some math and <i>HTML</i>: \(x^2\) and $$\frac{x+1}{x-1}$$",
    placeholder='Some HTML',
    description='Some HTML',
)
```

### Image

```python
# Yerel diskten veya bellekten okunan ham bayt (binary) verilerini Jupyter arayüzünde görsel/resim olarak görüntülemek ve boyutlandırmak (width, height) için kullanılır.
file = open("images/WidgetArch.png", "rb")
image = file.read()
widgets.Image(
    value=image,
    format='png',
    width=300,
    height=400,
)
```

## Button

```python
# ToggleButton gibi durum (True/False) saklamak yerine, tıklandığında bir Python fonksiyonunu tetiklemek (on_click) için kullanılan standart eylem butonudur; button_style, tooltip ve icon parametreleriyle özelleştirilebilir.
widgets.Button(
    description='Click me',
    disabled=False,
    button_style='', # 'success', 'info', 'warning', 'danger' or ''
    tooltip='Click me',
    icon='check'
)
```

```python
# Gerekli modülleri içe aktarma
import ipywidgets as widgets
from IPython.display import display

# Tam sayı kaydırma çubuğu oluşturup ekranda gösterme
w = widgets.IntSlider()
display(w)

# w.layout nitelikleri üzerinden widget'ın CSS/düzen özelliklerini dinamik olarak değiştirme:
# margin='auto': Widget'ı yatayda ortalar (boşlukları eşitler).
# height='75px': Widget bileşeninin yüksekliğini 75 piksel yapar.
w.layout.margin = 'auto'
w.layout.height = '75px'
```

```python
# Yeni bir IntSlider oluşturup ekranda gösterme
x = widgets.IntSlider(value=15, description='New Slider')
display(x)

# Layout nesnesini doğrudan aktarma / paylaşma:
# 'w' widget'ının layout özelliklerini (margin='auto', height='75px') birebir 'x' widget'ına kopyalar
# Böylece her iki widget aynı düzen ve boyutlandırma stilini paylaşır
x.layout = w.layout
```

---

```python
# widgets.Button: Ekranda buton bileşeni oluşturur.
# button_style='info': Butona hazır Bootstrap renk temasını uygular (açık mavi).
# Diğer alabileceği değerler: 'primary' (koyu mavi), 'success' (yeşil), 'warning' (turuncu/sarı), 'danger' (kırmızı) veya '' (varsayılan gri).
widgets.Button(description='Ordinary Button', button_style='info')
```

---

```python
# b1.style.button_color: Hazır Bootstrap temaları yerine CSS renk ismi ('lightgreen'), HEX kodu ('#90EE90') veya RGB değeriyle butona özel arka plan rengi tanımlar.
b1 = widgets.Button(description='Custom Color')
b1.style.button_color = 'lightgreen'
b1
```

```python
# b1.style.keys: İlgili widget'ın .style niteliği üzerinden doğrudan özelleştirilebilecek tüm geçerli CSS/stil anahtarlarını (button_color, font_weight vb.) liste olarak döndürür.
b1.style.keys
```

```python
# b2.style = b1.style: 'b1' butonuna uygulanan tüm özel stil ayarlarını (örneğin 'lightgreen' rengini) 'b2' butonuna doğrudan kopyalar
# Açıklama (description) belirtilmediği için buton üzerinde metin olmadan sadece renkli alan olarak görüntülenir
b2 = widgets.Button()
b2.style = b1.style
b2
```

```python
# description belirtmiş hali:
b2 = widgets.Button(description = 'New')
b2.style = b1.style
b2
```

---

```python
# s1.style.handle_color: IntSlider üzerindeki tutamaç (yuvarlak kaydırıcı düğmesi) rengini özelleştirir (örneğin 'blue').
s1 = widgets.IntSlider(description='My Handle')
s1.style.handle_color = 'blue'
s1
```

```python
# Slider bileşeninin .style niteliği altında değiştirilebilecek tüm geçerli stil anahtarlarını (handle_color vb.) listeler.
# buraya bakarak yukarda deneme yaptık 'handle.color' özelliğini kullandık
s1.style.keys
```

# Widget style traits

These are traits that belong to some of the more common widgets:

**Button**

* `button_color`
* `font_weight`

**IntSlider, FloatSlider, IntRangeSlider, FloatRangeSlider**

* `description_width`
* `handle_color`

**IntProgress, FloatProgress**

* `bar_color`
* `description_width`

Most others such as ToggleButton, Checkbox, Dropdown, RadioButtons, Select and Text only have `description_width` as an adjustable trait.

# GUI (Graphical User Interface) ve Jupyter Widgets Notları

## GUI Nedir?
**GUI (Graphical User Interface - Grafiksel Kullanıcı Arayüzü)**, programları sadece komut satırından (CLI) metin girerek çalıştırmak yerine; pencereler, butonlar, açılır menüler, metin kutuları ve kaydırma çubukları (slider) gibi görsel bileşenlerle kontrol etmeyi sağlayan arayüz yapısıdır.

---

## Python'da GUI Ekosistemi ve Kullanım Alanları

* **Geleneksel Masaüstü Arayüzleri:** `Tkinter`, `PyQt` veya `wxPython` gibi kütüphaneler işletim sistemi penceresi içinde çalışan bağımsız masaüstü uygulamaları geliştirmek için kullanılır.
* **Oyun Geliştirme:** 2D grafik çizimleri, pencere yönetimi ve oyun mekanikleri için doğrudan `PyGame` tercih edilir.
* **Web Uygulamaları:** Web projelerinde arayüz (Frontend) `HTML`, `CSS` ve `JavaScript` ile tasarlanırken, Python tarafı `Flask` veya `Django` çatıları kullanılarak sadece arka plan (Backend/API) mantığını yürütür.
* **Veri Analitiği ve Dashboard'lar:** Jupyter Notebook ortamında dış bir pencere açmak yerine, doğrudan notebook hücresi içinde etkileşimli paneller oluşturmak için **`ipywidgets`** kullanılır.

---

## Bu Bölümde İşlenecek Konular

* **`interact` Fonksiyonu:** Python fonksiyonlarına otomatik olarak slider, onay kutusu veya metin alanı bağlayarak parametre değişimlerini ve çıktıları anlık olarak görselleştirir.
* **Widget Temelleri (`ipywidgets`):** Buton (`Button`), tam sayı/ondalık kaydırıcılar (`IntSlider`, `FloatSlider`), seçim kutusu (`Checkbox`) ve açılır menü (`Dropdown`) bileşenlerinin manuel oluşturulması.
* **Olay Yönetimi (Event Handling):** Butona tıklama veya slider kaydırma gibi kullanıcı aksiyonlarını Python fonksiyonlarına bağlama (`on_click`, `observe`).
* **Stil ve Düzenleme (Styling & Layouts):** Arayüz elemanlarını yatay (`HBox`) ve dikey (`VBox`) hizalama, CSS benzeri stil tanımlamaları (`Layout`, `style`).
* **Etkileşimli Mini Panel (Dashboard):** Hücre kodlarını yeniden çalıştırmadan parametrelerin dinamik yönetildiği uçtan uca interaktif uygulama demosu.

# ipywidgets Stil ve Düzen (Styling & Layout) Rehberi

`ipywidgets` kütüphanesinde bileşenleri özelleştirmek için 3 temel mekanizma kullanılır:
1. **`button_style` / Önceden Tanımlı Temalar (Bootstrap)**
2. **`style` Parametresi (Bileşene Özel Ayrıntılar)**
3. **`layout` Parametresi (CSS / Boyutlandırma ve Konumlandırma)**

---

## 1. Hazır Temalar (`button_style` & `bar_style`)
Butonlar ve ilerleme çubukları (`IntProgress`) için Bootstrap renk şeması kullanılır:

* `''` (Varsayılan): Standart gri / renksiz
* `'primary'`: Koyu mavi (ana işlem)
* `'success'`: Yeşil (başarılı / onaylama)
* `'info'`: Açık mavi (bilgilendirme)
* `'warning'`: Sarı / turuncu (uyarı)
* `'danger'`: Kırmızı (silme / hata)

```python
import ipywidgets as widgets

# Buton üzerinde kullanımı
btn_primary = widgets.Button(description='Primary', button_style='primary')
btn_success = widgets.Button(description='Success', button_style='success')
btn_info = widgets.Button(description='Info', button_style='info')
btn_warning = widgets.Button(description='Warning', button_style='warning')
btn_danger = widgets.Button(description='Danger', button_style='danger')

# İlerleme çubuğunda kullanımı
progress = widgets.IntProgress(value=70, min=0, max=100, bar_style='success')

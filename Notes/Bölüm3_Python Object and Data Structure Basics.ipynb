### Indexing and Slicing With Strings
Slicingden kastı içindeki texti istediğimiz şekilde ayırıp alma,kullanma

```python
mystring = "Hello World"

print(mystring)
print(mystring[0])
print(mystring[-1])
```

```python
mystring = "abcdefghijk"
print(mystring)
print(mystring[2:]) #2.indekstan başlayıp sona kadar alır
print(mystring[:5]) #baştan başlayıp 5.indekse kadar alır
print(mystring[3:6]) 
#başlangıç indeksi 0, o yüzden başlangıç ve sonda 1 eksiğini düşünüyorsun sondan bölerken sonuncuyu dahil etmeyeceği için bir sonraki elemanı alman lazım
print(mystring[::2]) #3.parametre olan step size, kaçar kaçar atlayacağını belirler
```

```python
print(mystring[::-1]) #reverse (trick question on interviews)
```

### Reading, Writing, Appending Modes

* __mode="r"__ is read only
* __mode="w"__ is write only (will overwrite files or create new!)
* __mode="a"__ is append only (will ad on to files)
* __mode="r+"__ is reading and writing
* __mode="w+"__ is writing and reading (overwrites existing files or creates a new file)

```python
%%writefile my_new_file.txt
One on first
two on second
three on third
```

```python
with open("my_new_file.txt",mode="r") as f:
    print(f.read())
```

```python
with open("my_new_file.txt",mode="a") as f:
    f.write("four on fourth")
```

```python
with open("my_new_file.txt",mode="r") as f:
    print(f.read())
```

```python
with open("ajsdkasd.txt",mode="w") as f:
    f.write("I created this file") #even it's not exist, w command will create new file 
```

```python
with open("ajsdkasd.txt", mode="r") as f:
    print(f.read())
```

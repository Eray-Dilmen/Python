## Generator Nedir?
  * Normal fonksiyonlar (`return`) tüm listeyi bir anda hafızada (RAM) oluşturup toplu halde döner. 1 milyon eleman varsa 1 milyonluk yer kaplar.
  * Generator (`yield`) değerleri önceden üretip saklamaz. İhtiyaç duyuldukça **anlık olarak tek tek üretir**.

* **`yield` Ne Yapar?**
  * `yield`, fonksiyonu geçici olarak **duraklatır (pause)** ve o an üretilen değeri dışarı fırlatır.
  * Fonksiyon kaldığı satırı, döngüdeki indeksini ve yerel değişkenlerini hafızasında hatırlar.
  * Bir sonraki değer istendiğinde (örneğin `next()` çağrıldığında veya `for` döngüsünün yeni adımında), fonksiyon baştan başlamaz; tam durduğu yerden devam edip bir sonraki `yield`'e kadar çalışır.

* **`yield` Kısmına Ne Yazılır?**
  * O adımda dışarıya aktarmak istediğin nihai hesaplamayı/değeri yazarsın. Görseldeki `yield x**3` ifadesi: "Sıradaki küp değerini üret, çağırana ver ve bir sonraki istek gelene kadar burada bekle" anlamına gelir.

```python
def create_cubes(n):
    for x in range(n):
        yield x**3

# 1. Kullanım: for döngüsü ile tek tek tüketme (Bellek dostu)
for cube in create_cubes(5):
    print(cube)
```

```python
# 2. Kullanım: next() fonksiyonu ile adım adım değer üretme
cube_gen = create_cubes(3)
print(next(cube_gen))  # 0
print(next(cube_gen))  # 1
print(next(cube_gen))  # 8
```

```python
# 3. Kullanım: Tüm değerleri toplu olarak listeye dönüştürme
cubes_list = list(create_cubes(10))
print(cubes_list)
```

## Fibonacci Örneği

Fibonacci dizisi 1, 1, 2, 3, 5, 8, . . . şeklinde ilerler ve mantığı şudur: **Dizinin sıradaki elemanı her zaman `a` değişkeninde tutulur.**

Fonksiyonun içindeki adım adım işleyiş şöyledir:

* `yield a` **Satırı:** O anki Fibonacci sayısını (yani `a`'yı) dışarı fırlatır ve fonksiyonu orada duraklatır.
* `a, b = b, a + b` **Satırı:** Bir sonraki adıma hazırlık yapar; `b`'nin değerini `a`'ya aktarır, yeni toplamı da `b`'ye atar.

```python
def gen_fibon(n):
    a = 1
    b = 1
    for i in range(n):
        yield a
        a, b = b, a + b
```

```python
for number in gen_fibon(10):
    print(number)
```

**Adım Adım Değerler:**

1. **Adım 1:** `a = 1, b = 1` &rarr; `yield a` dışarı **1** verir. Sonra `a = 1, b = 2` olur.
2. **Adım 2:** `a = 1, b = 2` &rarr; `yield a` dışarı **1** verir. Sonra `a = 2, b = 3` olur.
3. **Adım 3:** `a = 2, b = 3` &rarr; `yield a` dışarı **2** verir. Sonra `a = 3, b = 5` olur.
4. **Adım 4:** `a = 3, b = 5` &rarr; `yield a` dışarı **3** verir. Sonra `a = 5, b = 8` olur.
5. **Adım 5:** `a = 5, b = 8` &rarr; `yield a` dışarı **5** verir. Sonra `a = 8, b = 13` olur.
6. **Adım 6:** `a = 8, b = 13` &rarr; `yield a` dışarı **8** verir. Sonra `a = 13, b = 21` olur.
7. **Adım 7:** `a = 13, b = 21` &rarr; `yield a` dışarı **13** verir. Sonra `a = 21, b = 34` olur.
8. **Adım 8:** `a = 21, b = 34` &rarr; `yield a` dışarı **21** verir. Sonra `a = 34, b = 55` olur.
9. **Adım 9:** `a = 34, b = 55` &rarr; `yield a` dışarı **34** verir. Sonra `a = 55, b = 89` olur.
10. **Adım 10:** `a = 55, b = 89` &rarr; `yield a` dışarı **55** verir. Sonra `a = 89, b = 144` olur.

`yield a` denmesinin sebebi; o anki sırası gelen sayının her zaman `a` kutusunun içinde hazır beklemesidir.

```python
print(list(gen_fibon(10)))
```

```python
g = gen_fibon(10)

print(next(g))  # 1
print(next(g))  # 1
print(next(g))  # 2
print(next(g))  # 3
print(next(g))  # 5
print(next(g))  # 8
print(next(g))  # 13
print(next(g))  # 21
print(next(g))  # 34
print(next(g))  # 55
```

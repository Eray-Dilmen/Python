### Class Object Attributes

Class'taki her instance için aynı olan attribute'lardır. (Değişmez)
Mesela Dog class'ından oluşturacağımız her objenin değişmeyeceği şey köpeğin bir 'memeli' oluşudur. 
O yüzden bunu '__ __init____' fonksiyonunun üstünde genel olarak tanımlarız (self yazmayız) 
species = 'mammal'

ve ne zaman bu class'tan bir obje üretsek attribute'larında species'i görebiliriz ki bunun value'si de direkt olarak 'mammal' olarak çıkar



```python
class Dog():
    species = 'mammal'
    
    def __init__(self,breed): 
        self.breed = breed
```

```python
my_dog = Dog(breed = 'Huskie')
my_dog.species
```

### Methods ( Operations / Actions )

```python
class Dog(): # class name'leri Camel Uppercase kullanıyoruz (kelime kelime DogClass gibi)
    def __init__(self,breed):
        self.breed = breed
    
    def bark(self,number): # methodları lowercase kullanıyoruz 
        print(f'WOOF! My breed is {self.breed}, I can bark {number} times ') # veya ''.format(self.breed) 
# görüldüğü üzere eğer bark fonksiyonuna da parametre atarsak {number} kısmında self dememize gerek yok. 

my_dog = Dog('Huskie')
my_dog.bark(7) # attribute aksine bunda çağırırken () kullanıyoruz
              #çünkü bu bir fonksiyon
```

```python
class Circle():
    # CLASS OBJECT ATTRIBUTE (same for every instance)
    pi = 3.14

    def __init__(self,radius=1):
        self.radius = radius

    def get_circumference(self):
        return self.radius * Circle.pi * 2 # Class Object olduğu için Circle.pi yazabiliyoruz, referans olarak veriyor. Veya onun yerine self.pi de yazabilirdik. İlki daha anlaşılır, genel oluyor.

my_circle = Circle(30)
my_circle.get_circumference() #fonksiyon olduğu için () ile çağırıyoruz
```

```python
print(my_circle.radius) # ama attribute'larını çağırırken () kullanmıyoruz 
```

## Inheritance and Polymorphism 


### inheritance 

```python
class Animal():
    def __init__(self):
        print("Animal Created")

    def who_am_i(self):
        print("I'm an animal")

    def eat(self):
        print("I'm eating")


## üstteki Animal nesnesini parametre olarak class'a vererek ordaki fonksiyonları miras almasını sağlıyoruz
## yani bu demek oluyor ki Dog nesnesinden türetilecek bir obje 'eat' fonksiyonunu kullanabilir.

class Dog(Animal):
    def __init__(self):
        Animal.__init__(self) # bu kısımda yukarıdaki constructordaki özelliklerin buraya taşınmasını sağlıyor
        # mesela self.a = a demek gibi, bu özellikleri de burda kullanmak istiyorsak bunu yazmamız lazım. Sadece özellik değil mantıksal karşılaştırmalar vs de olabilir her nesneye ayrı ayrı yapıştırmaktansa bu şekilde kullanıp otomatikleştirmek kod tekrarı açısından da büyük bir yükten kurtulmuş olunur.
        print("Dog Created")
        
    def bark(self):
        print("Woof!")

my_dog1 = Dog()
```

```python
my_dog1.eat() ## görüldüğü üzere, Animal nesnesinin fonksiyonu çalışıyor ve içindeki print ifadesi çalışıyor.
```

```python
## fakat eat fonksiyonunu da Dog nesnesinin içerisinde override edebiliriz.
class Dog(Animal):
    def __init__(self):
        print("Dog Created")

    def eat(self):
        print("Dog is eating")
        
    def bark(self):
        print("Woof!")

my_dog1 = Dog()
my_dog1.eat()
```

### Polymorphism (Çeşitlilik)

Farklı sınıflara ait nesnelerin aynı isimdeki bir metodu kendilerine özel biçimde çalıştırabilmesidir.

Örnek: Dog ve Cat sınıflarının ikisinde de speak() metodu vardır. 

dog.speak() çağrısı "Hav", cat.speak() çağrısı "Miyav" çıktısı verir.

```python
class Dog():

    def __init__(self, name):
        self.name = name

    def speak(self):
        return self.name + " says woof!"


class Cat():

    def __init__(self, name):
        self.name = name

    def speak(self):
        return self.name + " says meow!"

niko = Dog("niko")
felix = Cat("felix")
```

```python
for pet in [niko,felix]:
    print(type(pet))
    print(pet.speak() + '\n')
```

*çok biçimlilik, üstteki döngü örneğinde olduğu gibi farklı nesnelerin aynı metot çağrısına kendi tarzlarında yanıt vermesidir.*

---
#### başka bir örnek ( kullanım alanı )

```python
def pet_speak(pet):
    print(pet.speak())
```

```python
pet_speak(niko)
```

```python
pet_speak(felix)
```

### Abstract mantığı 

```python
class Animal():

    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError(
            "Subclass must implement this abstract method"
        )

myanimal = Animal('Den')
myanimal.speak()
```

## Hata vermesinin sebebi 
metodun içinde bilerek raise NotImplementedError(...) yazılmış olmasıdır. Bu bir kod hatası değil, bilerek fırlatılan bir istisnadır (exception).

## Mantığı şudur:

Animal sınıfı soyut (abstract) bir sınıf gibi tasarlanmış. 

"Genel bir Animal nesnesi doğrudan ses çıkaramaz, speak() metodunu sadece Dog veya Cat gibi alt sınıflar override ederek (ezerek) kullanmalıdır" mesajı verilmek isteniyor.

myanimal = Animal('fred') ile doğrudan ana sınıftan bir nesne ürettin.
myanimal.speak() dediğinde Python, Animal içindeki speak metodunu çalıştırdı.

Metodun içindeki raise NotImplementedError satırı devreye girdi ve sana "Bu metot henüz uygulanmadı, alt sınıfında bunu doldurman lazım" uyarısı verdi.

Yani: Animal sınıfından doğrudan speak() çağıramazsın; Dog sınıfı oluşturup speak() metodunu override etmen gerekir.

```python
class Cat(Animal):
    def speak(self):
        return self.name + " says meow!"
cake = Cat('Cake')
cake.speak()
```

### Special Methods // Dunder Methods // Magical Methods

normal şekilde print veya len gibi fonksiyonal özellikleri Class'lar için kullanamıyoruz 

Onlar için özel metodlar kullanıyoruz (O da class'ın içine gömmek)

Class'ın içine special method olarak adlandırılan özel isimli fonksiyonlar yaratıyoruz ve dışardan fonksiyonlara classtan üretilen objeyi verdiğimizde 
Class'ın içerisindeki özel methodlar çalışıyor ve içine yazdığımız şeyler gerçekleşiyor.


```python
class Book():
    def __init__(self, author, title, pages):
        self.author = author
        self.title = title
        self.pages = pages 
        
    def __str__(self):
        return f"{self.title} has {self.pages} pages made by {self.author}"

    def __del__(self):
        print(f"{self.title} book has been deleted")

    def __len__(self):
        return self.pages
```

```python
# def__str__ olmadan önceki hali:
book2 = Book("Eray","Fantastik Maceralar",576)
print(book2)
```

```python
#bellekteki yerini gösteriyor çünkü book2'nin referans olarak gösterebileceği bir str yok.
# javada da oop öğrenirken bu tarz bir şey vardı class içerisinde string fonksiyonu tanımlıyorduk
```

```python
# def__str__ olduktan sonraki hali:
book2 = Book("Eray","Fantastik Maceralar",576)
print(book2)
```

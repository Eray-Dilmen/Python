```python
s1 = {1,2,3}
s2 = {2,3,4}

s1.intersection(s2) # kesişimini alıp döndürüyor
```

```python
s1 = {1,2}
s2 = {2,3,4}
s3 = {5}

s1.isdisjoint(s2) # kesişimi var mı ? false dönecek çünkü 2 var
```

```python
s1.isdisjoint(s3)
```

```python
s1 = {1,2}
s2 = {1,2,4}
s1.issubset(s2) # s1 s2'nin alt kümesi mi ? 
```

```python
s2.issuperset(s1) # s2 s1'i kapsıyor mu ? (s2 s1'in üst kümesi mi ?)
```

```python
s1 = {1,2}
s2 = {1,2,4}

s1.symmetric_difference(s2) # ikisinde ortak olmayan elemanlar (kesişimin tersi)
```

```python
s1.union(s2) # birleşimini return ediyor. Değişkene atanabilir
```

```python
s1
```

```python
s1.update(s2)
s1 # unionla farkı bu orjinal kümeyi 'update' ediyor
```

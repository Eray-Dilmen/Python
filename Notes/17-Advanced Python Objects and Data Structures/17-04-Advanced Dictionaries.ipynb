# Advanced Dictionaries

```python
d = {'k1': 1, 'k2': 2}
```

```python
{k:v**2 for k,v in d.items()}
```

```python
{x:x**2 for x in range(10)}
```

```python
for k in d.keys():
    print(k)
```

```python
for k in d.values():
    print(k)
```

```python
for k in d.items():
    print(k)
```

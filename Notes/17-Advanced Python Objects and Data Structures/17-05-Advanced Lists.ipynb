# append ve extend farkı 

```python
x = [1,2,3]
x.append([4,5]) # Verilen nesneyi tek bir eleman olarak listenin sonuna ekler
x               #(iç içe liste / nested list oluşturur)
```

```python
x = [1,2,3]
x.extend([4,5]) # Verilen dizinin (iterable) elemanlarını tek tek çıkarıp mevcut listeye katar
x               #(listeleri uç uca birleştirir)
```

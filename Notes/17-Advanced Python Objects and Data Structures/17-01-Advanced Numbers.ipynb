```python
round(3.14813923, 2) # Virgülden sonraki (ondalık) n. basamağa göre yuvarlar (örn: round(3.1428, 2) → 3.14).
# ilk 2 basamağını alırken 3.basamağına bakar 5 ve üstüyse 2.basamağı 1 yükseltir.
```

```python
# altında ise sabit tutar 
round(3.14313923, 2)
```

```python
round(33.14283923, -1) # negatif girince virgülden önceki basamaklara gider:
# -1: En yakın 10'un katına yuvarlar (3.14→0.0, 16.5→20.0).
# -2: En yakın 100'ün katına yuvarlar (140→100, 180→200).
# -3: En yakın 1000'in katına yuvarlar (1250→1000).
```

```python
round(155.12834, -1) # 10 un katına yukarı yuvarlayacak çünkü son basamak 5
```

```python
round(155.12834, -2) # 100'ün katında yukarı yuvarlayacak çünkü 55 
```

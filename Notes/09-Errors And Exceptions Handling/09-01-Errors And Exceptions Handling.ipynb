### Exception Handling
Hatayı yönetme biçimidir. sistemin çökmemesi veya durmaması için çalışırken aldığımız hataları sistem çökmeden sorunsuz bir şekilde hallederek yönetmemizi ve sistemin çalışmasını sürdürmesini sağlar. 

```python
try:
    result = 10 + 10
except: # bu genel bir error şeklidir, her türlü error'u yakalar ve bu kod bloğunu çalıştırır, aşağıda biraz daha farklısını göreceğiz
    print("You seem try to sum int with some other type, please try again")
```

```python
# hatasız çalıştı çünkü 10 ile 10 integer ile integer'in toplamıdır, şimdi hatalı şekilde toplama yapalım:
```

```python
try:
    result = 10 + '10'
except:
    print("You seem try to sum int with some other type, please try again")
```

```python
# şimdi spesifik olarak bir error oluştuğunda o error olursa ne olacağını seçelim.
# Birden fazla error çeşidi için farklı kod blokları oluşturup farklı şeyler çalıştırabiliriz.
```

```python
try:
    result = 10 + '10'
except TypeError: 
    print("You seem enter type wrong, please try again")

except ZeroDivisionError:
    print("Can't divide into 0, please try again")
```

```python
# sadece TypeError çalıştı çünkü int ile string toplamaya çalıştık. 
```

```python
try:
    result = 10 + 10
    division = 10/0
except TypeError: 
    print("You seem enter type wrong, please try again")

except ZeroDivisionError:
    print("Can't divide into 0, please try again")
```

```python
# 0'a bölünme hatası aldık. 
```

## Finally 
Finally hata olup olmaması farketmeksizin içine yazılan kod bloğunun çalışmasını sağlar

```python
try:
    result = 10 + 10
    division = 10/0
except TypeError: 
    print("You seem enter type wrong, please try again")

except ZeroDivisionError:
    print("Can't divide into 0, please try again")
finally:
    print("Executed regardless errors")
```

```python
#görüldüğü gibi hata olsa bile o kod bloğu çalıştı
```

```python
def ask_for_int():
    try:
        result = int(input("Please provide number: "))
    except:
        print("You didn't enter number, please try again")
    finally:
        print("End of the try/except/finally")

ask_for_int()
```

```python
## except yoksa ne olacağını çalıştıran kod kısmı 'else'
## while'da True kullanmışsak bir yerlere döngüden çıkmak için break koymamız lazım
```

```python
def ask_for_int():
    while True:
        try:
            result = int(input("Please provide number: "))
        except:
            print("You didn't enter number, please try again ")
            continue
        else:
            print("Your number is taken")
            break
        finally:
            print("End of the try/except/finally")

ask_for_int()
```

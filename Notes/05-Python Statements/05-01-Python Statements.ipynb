### FOR LOOPS

```python
mylist = [1,2,3,4,5];

for numb in mylist: 
    print(numb)
```

```python
for numb in mylist:
    print("hello") #5 kere basmasını sağlayacak çünkü iterable,
                   #objeyi içindeki kadar geziyoruz
```

```python
for n in mylist:
    if n%2 == 0:
        print (n)
        print("\n")
    else:
         print(f'odd numb: {n}')
         print("print 2.yol: ", n)
         print("\n")
```

### Tuples unpacking
---

```python
 mylist = [(1,2,3),(5,6,7),(9,10,11)]

for a,b,c in mylist: 
    print("\n",b)
    print(a,c)
```

## küçük bir örnek

```python
sayilar = [1, 2, 3, 4]
karesi = [num**2 for num in sayilar]# Çıktı: [1, 4, 9, 16]
karesi
```

```python
sayilar = [1, 2, 3, 4, 5, 6]
ciftler = [x for x in sayilar if x % 2 == 0]  # Çıktı: [2, 4, 6]
ciftler
```
